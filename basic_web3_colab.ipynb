## 2) Connect to Ethereum (public RPC)

```python
from web3 import Web3

RPC_URL = "https://cloudflare-eth.com"
w3 = Web3(Web3.HTTPProvider(RPC_URL))

print('Connected:', w3.is_connected())
print('Chain ID:', w3.eth.chain_id)
```

## 3) Read latest block data

```python
latest_block_number = w3.eth.block_number
latest_block = w3.eth.get_block(latest_block_number)

print('Latest block number:', latest_block_number)
print('Block hash:', latest_block.hash.hex())
print('Transaction count:', len(latest_block.transactions))
```

## 4) Check account balance

```python
# Example address (Ethereum Foundation wallet)
address = w3.to_checksum_address('0xde0b295669a9fd93d5f28d9ec85e40f4cb697bae')

balance_wei = w3.eth.get_balance(address)
balance_eth = w3.from_wei(balance_wei, 'ether')

print('Address:', address)
print('Balance (ETH):', balance_eth)
```
